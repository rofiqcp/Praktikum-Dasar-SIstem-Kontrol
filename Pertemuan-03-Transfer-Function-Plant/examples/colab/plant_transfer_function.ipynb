# Transfer Function Plants
Colab companion for water heater and DC motor.


```python
!pip -q install control
import control as ct, numpy as np, matplotlib.pyplot as plt
```

```python
s=ct.TransferFunction.s
Gheater=0.8/(120*s+1)
R,L,K,J,b=2,0.5,0.1,0.02,0.2
Gspeed=K/((J*s+b)*(L*s+R)+K*K)
Gpos=Gspeed/s
for name,G,T in [('heater',Gheater,np.linspace(0,700,1000)),('speed',Gspeed,np.linspace(0,5,1000)),('position',Gpos,np.linspace(0,5,1000))]:
    t,y=ct.step_response(G,T); plt.figure(); plt.plot(t,y); plt.grid(); plt.title(name)
```
