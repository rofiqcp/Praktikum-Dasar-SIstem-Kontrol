```python
!pip -q install control
import control as ct, numpy as np, matplotlib.pyplot as plt
```

```python
s=ct.TransferFunction.s
Gt=0.8/(45*s+1)
J,b,K,R,L=0.01,0.1,0.01,1,0.5
Gw=K/((J*s+b)*(L*s+R)+K*K)
Gtheta=Gw/s
for name,G in [('heater',Gt),('speed',Gw),('position',Gtheta)]:
    t,y=ct.step_response(G)
    plt.figure(); plt.plot(t,y); plt.grid(); plt.title(name); plt.xlabel('s')
plt.show()
```
